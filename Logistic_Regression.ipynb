```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
data = {
    "Hours_Studied": [1,2,3,4,5,6,7,8,2,6,7,3],
    "Attendance":    [50,55,60,65,70,80,85,90,40,75,88,58],
    "Result":        [0,0,0,0,1,1,1,1,0,1,1,0]  # 0=Fail, 1=Pass
}
```

```python
df = pd.DataFrame(data)
display(df)
```

```python
plt.scatter(df['Hours_Studied'],df['Attendance'],c=df['Result'])
plt.xlabel('Hours_Studied')
plt.ylabel('Attendance')
plt.title("pass(1) vs fail(0)")
plt.show()
```

```python
X = df[["Hours_Studied","Attendance"]]
Y = df["Result"]
```

```python
display(X)
display(Y)
```

```python
model = LogisticRegression()
model.fit(X,Y)
```

```python
predictions = model.predict(X)
predictions
type(predictions)
```

```python
new_student = pd.DataFrame({
    "Hours_Studied":[4],
    "Attendance":[85]
})

model.predict(new_student)
```

```python
print(model.classes_)
model.predict_proba(new_student)
```

```python
accuracy = accuracy_score(Y, predictions)
print("Accuracy:", accuracy)
```

```python
new_student = pd.DataFrame({
    "Hours_Studied":[3],
    "Attendance":[90]
})

model.predict(new_student)
```

```python
cm = confusion_matrix(Y, predictions)
print(cm)
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(X.values, Y.values, clf=model)
plt.xlabel("Hours Studied")
plt.ylabel("Attendance")
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
import seaborn as sns
```

```python
cancer_data = load_breast_cancer()
display(cancer_data)
```

```python
X= pd.DataFrame(data = cancer_data.data, columns = cancer_data.feature_names)
Y = cancer_data.target
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
display(X)
display(Y)
```

```python
from sklearn.preprocessing import StandardScaler

# Instantiate the scaler
scalar = StandardScaler()

# Fit on training data and transform it
X_train_scaled = scalar.fit_transform(X_train)

# Transform testing data using the fitted scaler (to prevent leakage)
X_test_scaled = scalar.transform(X_test)
```

```python
model = LogisticRegression()
model.fit(X_train_scaled,Y_train)
```

```python
from sklearn.metrics import classification_report
Y_pred = model.predict(X_test_scaled)
acc = accuracy_score(Y_test,Y_pred)
print(f"the accuracy:{round((acc*100),2)}%")
report = classification_report(Y_test,Y_pred, target_names= ["Malignant(0)","Benign(1)"])
print(report)
cm = confusion_matrix(Y_test, Y_pred)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot = True, cmap = "Reds",
            xticklabels=["Predicted Malignant(0)","Predicted Benign(1)"],
            yticklabels=["Actual Malignant","Actual Benign"])
plt.show()

```
